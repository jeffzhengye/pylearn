# chameleon

* https://www.toutiao.com/article/7587611587171631657/?log_from=56902e628d0b1_1767544110217
* https://github.com/malthe/chameleon

## 核心特性
极致性能：模板被编译为Python字节码，渲染速度接近原生Python
标准语法：采用TAL/TALES/METAL规范，语法成熟稳定
自动转义：默认对输出内容进行HTML转义，有效防止XSS攻击
模板继承：支持宏定义和模板复用，便于构建复杂页面结构
调试友好：错误信息精确定位到模板源码行号
双模式支持：同时支持HTML和XML两种模板格式
